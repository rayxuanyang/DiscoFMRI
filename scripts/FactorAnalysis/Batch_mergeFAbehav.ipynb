Merge the factor scores, 106 features, and all behavioral outcome measures. 

```python
import pandas as pd
import os
import seaborn as sns
import matplotlib as mpl
import matplotlib.pyplot as plt
import scipy.stats as stats
import numpy as np
from matplotlib.patches import Rectangle

Dir_github = '/work/desai-lab/xuanyang/Project/Semantic/dissemination/github/DiscoFMRI'
repo = 'FactorAnalysis_fMRI'
Dir_output = os.path.join(Dir_github,repo,'scripts','FactorAnalysis','output')
NFA = 8
```

```python
df_SCOPE_raw = pd.read_csv('/work/desai-lab/xuanyang/Project/Semantic/analysis/FactorAnalysis/github/data/SCOPE_word.csv')
df = pd.read_csv(os.path.join('/work/desai-lab/xuanyang/Project/Semantic/analysis/FactorAnalysis/github/data/SCOPE_word.csv'), usecols=["Word"], keep_default_na=False, na_values=[],encoding="latin-1")
df_SCOPE_raw['Word'] = df['Word'] 
df_SCOPE_raw = df_SCOPE_raw.loc[~df_SCOPE_raw['Word'].duplicated()].reset_index() # the word "used" is duplicated
```

```python
NFA = 8
df_FA_raw = pd.read_csv(os.path.join(Dir_output,f'data_FA_n{NFA}.csv'))
df_loadings = pd.read_csv(os.path.join(Dir_output,f'FA_n{NFA}_loadings.csv'))
# recode the FA7 to be sparse - dense
df_FA_raw['F7_n8_raw'] = df_FA_raw['F7_n8']
df_FA_raw['F7_n8'] = df_FA_raw['F7_n8']*-1

df_FA = df_FA_raw[['idx','Word','Lemma']+list(df_loadings['index'])].copy()
df_FA['FA7_raw'] = df_FA_raw['F7_n8_raw']
df_FA[[f'FA{x}' for x in range(1,NFA+1)]] = df_FA_raw[[f'F{x}_n{NFA}' for x in range(1,NFA+1)]]

List_106vars = list(df_loadings['index'])
List_FAs = [f'FA{x}' for x in range(1,NFA+1)]
df_FA['flag_FA'] = 1
```

```python
# add behavioral variables
List_var_RT = [x for x in df_SCOPE_raw.columns if '_RT_' in x and '_z' in x]
List_var_ACC = [x for x in df_SCOPE_raw.columns if '_ACC_' in x]
List_var_Y = List_var_RT+['Recog_Memory']+List_var_ACC

df_behav = df_SCOPE_raw[['Word']+List_var_Y].copy()

df_FA_behav = pd.merge(left=df_FA,right=df_behav,left_on = 'Word',right_on = 'Word')
df_FA_behav.to_csv(os.path.join(Dir_output,'df_FA8_behav_merged.csv'),index=False)
df_FA_behav
```
