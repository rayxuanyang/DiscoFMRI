```python
import matplotlib.pyplot as plt
import numpy as np
def generate_radar(sample,sig,filename):
    N = len(sample)
    
    # Angles for each bar
    theta = np.linspace(0, 2*np.pi, N, endpoint=False)
    width = 2*np.pi / N *.97 # Equal bar widths
    
    fig = plt.figure(figsize=(2, 2))
    ax = fig.add_subplot(111, polar=True)

    
    # Set transparent background
    fig.patch.set_alpha(0)  # Make the figure background transparent
#     ax.set_facecolor('none')  # Make the axis background transparent
    ax.set_facecolor('white')
    
    # Plot bars: red for positive, blue for negative
    for i in range(N):
        # color = '#731A73' if sample[i] >= 0 else '#F898CB'
#         color = 'red' if sample[i] >= 0 else 'blue'
        color = '#AC3C3E' if sample[i] >= 0 else '#286BBA'
        
        if sig[i] == 0:
            color = 'gray'
        ax.bar(theta[i], abs(sample[i]), width=width, color=color, alpha=0.7)
        
    ax.set_xticks(theta)
    
    ax.set_xticklabels([f'{x}' for x in range(1, 9)], ha='center', va='center',fontsize=15)
    ax.xaxis.set_tick_params(pad=-24)  # Adjust the pad to move labels inward
    ax.set_yticklabels([])  # Hide y-axis tick labels
    
    ax.set_rlim(0, 9)  # Adjust the upper limit as needed
    
    # ax.yaxis.set_tick_params(size=0)  # Hide y-axis tick marks

    ax.yaxis.grid(True)
    
    plt.close()
    fig.savefig(filename, transparent=True, dpi=300)
```

```python
Dir_output = os.path.join(Dir_results_2nd,'threshold_p005FDR','conjunction','radarplot')
if not os.path.exists(Dir_output):
    os.makedirs(Dir_output)
    
for i,row in df_conjunction.loc[df_conjunction['n']>=1].iterrows():
    ROI = row['ROI_idx']
    sample = df_results.loc[df_results['ROI_idx']==ROI,'ROI_value'].values
    sig = df_results.loc[df_results['ROI_idx']==ROI,'significant_FDR'].values
    
    
    filename = os.path.join(Dir_output,'byROIidx',f"radar_{int(ROI)}_{row['n']}FAs_{row['FA']}.png")
    os.makedirs(os.path.join(Dir_output,'byROIidx'),exist_ok=True)
    
    generate_radar(sample,sig,filename)

    filename = os.path.join(Dir_output,f"radar_{row['n']}FAs_{row['FA']}_{int(ROI)}.png")
    generate_radar(sample,sig,filename)
```
