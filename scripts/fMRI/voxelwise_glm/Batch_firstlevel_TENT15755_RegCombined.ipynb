```python
import os 
import pandas as pd
N_factors = 8
Dir_FA = os.path.join(Dir_github,project,'scripts','FactorAnalysis','output')
df_FA = pd.read_csv(os.path.join(Dir_FA,f'data_FA_n{N_factors}.csv'))
df_FA
```

```python
N_allwords = len(df_timestamps)
df_merge1 = df_timestamps.merge(df_FA, left_on="Word", right_on="Word", how="left")
df_merge2 = df_timestamps.merge(df_FA, left_on="word_gentle", right_on="Word", how="left")
df_merge12 = df_merge1.combine_first(df_merge2)
df_merge3 = df_timestamps.merge(df_FA, left_on="word_lower", right_on="Word", how="left")
df_merge123 = df_merge12.combine_first(df_merge3) # Use this one
df_merge4 = df_timestamps.merge(df_FA, left_on="lemma", right_on="Word", how="left")
df_merge1234 = df_merge123.combine_first(df_merge4) 

df_final = df_merge123.copy()
df_final
```

```python
List_vars_default = ['word','transcript','PoS','functional']
NFA=8
List_vars = [f'F{x}_n{NFA}' for x in range(1,NFA+1)]
df_final[List_vars_default+List_vars].dropna()
```

```python
df_table = df_final[['word','transcript']].dropna().groupby(['transcript']).count()
df_table['tokens_analyze'] = df_final[['F1_n8','onset','transcript']].dropna().groupby(['transcript']).count()['F1_n8']
df_table['tokens_analyze%'] = df_table['tokens_analyze']/df_table['word']*100
df_table['tokens_analyze%'] = df_table['tokens_analyze%'].round(2)
df_table.reset_index(inplace=True)
print('Milkyway: average')
print(df_table.loc[df_table['transcript'].str.contains('milkyway'),['word','tokens_analyze','tokens_analyze%']].mean())
df_table
```

```python
df_tokens_summary = pd.DataFrame()
for i,df,label in zip(range(5),[df_timestamps,df_merge1,df_merge12,df_merge123,df_merge1234],
                     ['raw','Word','Word_gentle','word_lower','lemma']):
    df_tokens_summary.loc[i,'label'] = label
    df_tokens_summary.loc[i,'tokens_all'] = df[['Word']+[x for x in df.columns if x.startswith('F') and '_n' in x]].dropna()['Word'].shape[0]
    df_tokens_summary.loc[i,'words_all'] = df[['Word']+[x for x in df.columns if x.startswith('F') and '_n' in x]].dropna()['Word'].unique().shape[0]
    df_tokens_summary.loc[i,'tokens_content'] = df[['Word']+[x for x in df.columns if x.startswith('F') and '_n' in x]].dropna().loc[df['functional']==0,'Word'].shape[0]
    df_tokens_summary.loc[i,'words_content'] = df[['Word']+[x for x in df.columns if x.startswith('F') and '_n' in x]].dropna().loc[df['functional']==0,'Word'].unique().shape[0]
    
#     print(f"all words: {N_all_tokens} tokens, {N_all_words} words")
#     print(f"content words: {N_content_tokens} tokens, {N_content_words} words")
for var in ['tokens_all','words_all','tokens_content','words_content']:
    df_tokens_summary[var+'%']  = df_tokens_summary[var]/df_tokens_summary.loc[0,var]
df_tokens_summary
```

```python
Dir_working = '/work/desai-lab/xuanyang/Project/Semantic/analysis/ParametricModulation/Nastase/allstories/FactorAnalysis'
Nvar = 113
NFA = 8
flag_model = f'Nvar{Nvar}NFA{NFA}_LPAC_multipleReg'

List_vars_default = ['word','onset','transcript','PoS','functional']

mask_PAC_L = '/work/desai-lab/xuanyang/Project/dataset/Nastase/codes/mask/final/rs_HCP_AuditoryMask_L_sm6_100.nii.gz'
mask_PAC_R = '/work/desai-lab/xuanyang/Project/dataset/Nastase/codes/mask/final/rs_HCP_AuditoryMask_R_sm6_100.nii.gz'

List_vars = [f'F{x}_n{NFA}' for x in range(1,NFA+1)]
    
Dir_model = os.path.join(Dir_working,'models',flag_model,f'FA{NFA}')
Dir_results = os.path.join(Dir_model,'results')
Dir_results_1st = os.path.join(Dir_results,'firstlevel')
Dir_results_2nd = os.path.join(Dir_results,'secondlevel')
if not os.path.exists(Dir_results):
    os.makedirs(Dir_results)
if not os.path.exists(Dir_results_1st):
    os.makedirs(Dir_results_1st)
    
Dir_scriptsToRun = os.path.join(Dir_model,'codes','scriptsToRun')
if not os.path.exists(Dir_scriptsToRun):
    os.makedirs(Dir_scriptsToRun)
Dir_log = os.path.join(Dir_model,'codes','logs')
if not os.path.exists(Dir_log):
    os.makedirs(Dir_log)
for i,subID, task,label in zip(df_master.index,df_master.subID,df_master.task,df_master.label):
    idx = "{:03d}".format(i+1)

    df_event = pd.read_csv(df_master.loc[i,'event'],sep='\t')
    blankSecs = df_event.loc[0,'onset']
    df_timestamps_tmp = df_final.loc[df_final['transcript']==label,List_vars_default+List_vars].reset_index(drop=True)
    df_timestamps_tmp['onset'] = df_timestamps_tmp['onset'] + blankSecs
    df_timestamps_tmp['missing'] = 1
    df_timestamps_tmp.loc[df_timestamps_tmp[List_vars].notna().all(axis=1),'missing'] = 0

    Dir_sub = os.path.join(Dir_results_1st,subID)
    if not os.path.exists(Dir_sub):
        os.makedirs(Dir_sub)
    Dir_sub_task = os.path.join(Dir_sub,task)
    if not os.path.exists(Dir_sub_task):
        os.makedirs(Dir_sub_task)

    ### Timings
    # missing
    df_tmp = df_timestamps_tmp.loc[df_timestamps_tmp.missing==1]
    df_tmp.to_csv(os.path.join(Dir_sub_task,'{}_{}_missing.csv'.format(subID,task)),index=False)
    List_output = df_tmp.onset.astype(str).values
    path_missing = os.path.join(Dir_sub_task,'{}_{}_missing.1D'.format(subID,task))
    with open(path_missing,'w') as f:
        f.writelines('\n'.join(List_output))

    # content words
    df_tmp = df_timestamps_tmp.loc[(df_timestamps_tmp.functional==0)&(df_timestamps_tmp.missing==0)]
    df_tmp.to_csv(os.path.join(Dir_sub_task,'{}_{}_cont.csv'.format(subID,task)),index=False)
    List_tmp = [','.join(x) for x in df_tmp[List_vars].astype(str).values]
    List_output =  [x+'*'+y for x,y in zip(df_tmp.onset.astype(str),List_tmp)] 
    path_cont = os.path.join(Dir_sub_task,'{}_{}_cont.1D'.format(subID,task))
    with open(path_cont,'w') as f:
        f.writelines('\n'.join(List_output))

    # functional words
    df_tmp = df_timestamps_tmp.loc[(df_timestamps_tmp.functional==1)&(df_timestamps_tmp.missing==0)]
    df_tmp.to_csv(os.path.join(Dir_sub_task,'{}_{}_func.csv'.format(subID,task)),index=False)
    List_output =  df_tmp.onset.astype(str).values
    path_func = os.path.join(Dir_sub_task,'{}_{}_func.1D'.format(subID,task))
    with open(path_func,'w') as f:
        f.writelines('\n'.join(List_output))


    ### Confounding variables
    path_confounds = df_master.loc[i,'confounding']
    df_confounds = pd.read_csv(path_confounds,sep='\t')
    df_motions = df_confounds[['trans_x','trans_y','trans_z','rot_x','rot_y','rot_z']].copy()
    path_motion = os.path.join(Dir_sub_task,'motions_6.1D')
    #if not os.path.exists(path_motion):
    df_motions.to_csv(path_motion,index=False,header=False,sep=' ')
    list_outlier = [outlier for outlier in df_confounds.columns if 'motion_outlier' in outlier]
    df_confounds['motion_outlier'] = 1-df_confounds[list_outlier].sum(axis=1)
    path_motion_outlier = os.path.join(Dir_sub_task,'motions_outlier.1D')
    df_confounds['motion_outlier'].to_csv(path_motion_outlier,index=False,header=False,sep=' ')

    
    img_smooth = df_master.loc[i,'smooth_clean_rs']
    scriptFile = os.path.join(Dir_sub_task,'Batch_firstlevel_{}_{}.sh'.format(subID,task))
    with open(scriptFile, "w") as f:
            f.write("#!/bin/tcsh\n")
            f.write("/work/apps/AFNI/linux_openmp_64/afni_proc.py -subj_id {} \\\n".format(subID))
            f.write("-scr_overwrite -blocks mask regress \\\n")
            f.write("-mask_import PAC_L {} \\\n".format(mask_PAC_L))
            f.write("-dsets {} \\\n".format(img_smooth))
            f.write("-regress_stim_times {} {} {} \\\n".format(path_cont,path_func,path_missing))
            f.write("-regress_basis 'TENT(1.5,7.5,5)' \\\n")
#                 f.write("-regress_ROI {} {} \\\n".format('PAC_L','PAC_R'))
            f.write("-regress_ROI {} \\\n".format('PAC_L'))
            f.write("-regress_stim_types AM2 times times \\\n")
            f.write("-regress_stim_labels cont func missing \\\n")
            f.write("-regress_opts_3dD -jobs 6 \\\n")
            for iFA in range(1,NFA+1):
                f.write(f"-gltsym 'SYM: cont[{1+5*iFA}] +cont[{2+5*iFA}] +cont[{3+5*iFA}] +cont[{4+5*iFA}]' -glt_label {iFA} FA{iFA}_bin2345 \\\n")
            f.write("-regress_polort 0 \\\n")
            f.write("-regress_3dD_stop \\\n")
            f.write("-regress_reml_exec \\\n")
            f.write("-regress_make_ideal_sum sum_ideal.1D \\\n")
            f.write("-regress_run_clustsim no \\\n")
            f.write("\n")

    masterScript = os.path.join(Dir_scriptsToRun,'run_Batch_firstlevel_{}.sh'.format(idx))
    with open(masterScript, "w") as f:
        f.write("#!/bin/tcsh\n")
        f.write("cd {} \n".format(Dir_sub_task))
        f.write("tcsh {} \n".format(scriptFile))
        f.write("tcsh proc.{} \n".format(subID))
        f.write("rm {}/{}.results/pb00.* \n".format(Dir_sub_task,subID))
        f.write("rm {}/{}.results/all_runs.* \n".format(Dir_sub_task,subID))
        f.write("rm {}/{}.results/fitts.* \n".format(Dir_sub_task,subID))
        f.write("rm {}/{}.results/errts.* \n".format(Dir_sub_task,subID))
    print(iFA,idx,subID)

Dir_codes = os.path.join(Dir_model,'codes')

if not os.path.exists(Dir_log):
    os.makedirs(Dir_log)

jobname = flag_model
ncores = 6
partition = 'desailab-48core'
path_output = os.path.join(Dir_log,jobname)
script_slurm = os.path.join(Dir_codes,'Batch_slurm_firstlevel.sh')
array = '001-213'
#array = '001-013'
with open(script_slurm, "w") as f:
    f.write("#!/bin/sh\n")
    f.write("### name your job \n")
    f.write("#SBATCH --job-name={} \n".format(jobname))
    f.write("### specify number of cores to use \n")
    f.write("#SBATCH -n {} \n".format(ncores))
    f.write("#SBATCH -N 1 \n".format(ncores))
    f.write("### which partition do you want to use? \n")
    f.write("#SBATCH -p {} \n".format(partition))
    f.write("### %a specifies that the output file for each subject will be named according to their array ID \n")
    f.write("# Where to output log files? \n")
    f.write("#SBATCH --output='{}-%A_%a.log' \n".format(path_output))
    f.write("#SBATCH --error='{}-%A_%a.err' \n".format(path_output))
    f.write("### list the subjects that you want to submit in this array \n")
    f.write("#SBATCH --array={} \n".format(array))
    f.write("module load afni \n")
    f.write("module load python3/anaconda/2021.11 \n")
    f.write('echo "Slurm job ID: " $SLURM_JOB_ID \n')
    f.write('echo "Slurm array task ID: " $SLURM_ARRAY_TASK_ID \n')

    f.write('printf -v subj "%03d" $SLURM_ARRAY_TASK_ID \n')
    f.write('echo "Running firstlevel regression on sub-$subj" \n')
    f.write('cd {} \n'.format(Dir_scriptsToRun))
    f.write('tcsh run_Batch_firstlevel_$subj.sh \n')
    f.write('echo "Finished firstlevel regression sub-$subj" \n')
    f.write('date \n')
```
