## Select stories from Narratives dataset for data analysis 

```python
import os
import pandas as pd
import json
Dir_project = '/work/desai-lab/xuanyang/Project/dataset/Nastase/narratives'
```

__[The original dataset description](https://www.ncbi.nlm.nih.gov/pmc/articles/PMC8479122/table/Tab1/?report=objectonly)__

#### Import the master sheet of participants

```python
df_participants_raw = pd.read_csv(os.path.join(Dir_project,'participants.tsv'),sep='\t')
for x in df_participants_raw.columns:
    df_participants_raw.loc[df_participants_raw[x].isna(),x] = 'n/a'

print(f"The Narratives fMRI dataset contains {df_participants_raw.shape[0]} participants in total.")
df_participants_raw
```

```python
df_participants = []
for i,row in df_participants_raw.iterrows():
    df_tmp = pd.DataFrame({'subID':row['participant_id'],'age':row['age'].split(','),
                           'sex':row['sex'].split(','),
                           'task':row['task'].split(','),
                           'condition':row['condition'].split(','),
                           'comprehension':row['comprehension'].split(',')})
    df_participants.append(df_tmp)
df_participants = pd.concat(df_participants,ignore_index=True)
df_participants['N'] = 1
print(f"The Narratives fMRI dataset contains {df_participants.shape[0]} scans from {df_participants_raw.shape[0]} participants in total.")
df_participants
```

#### Scans need to be excluded

```python
f = open(os.path.join(Dir_project,'code','scan_exclude.json'))
df_scan_exclude = []
dict_scan_exclude = json.load(f)
for task in dict_scan_exclude:
    if len(dict_scan_exclude[task])!=0:
        for subID in dict_scan_exclude[task]:
            df_tmp = pd.DataFrame({'subID':subID,'task':task},index=[0])
            df_scan_exclude.append(df_tmp)
df_scan_exclude = pd.concat(df_scan_exclude,ignore_index=True)
print(f'{len(df_scan_exclude)} scans from {len(df_scan_exclude.subID.unique())} participants need to be excluded.')
df_scan_exclude.sort_values(['subID'])
df_participants_filter = []
for i,row in df_participants.iterrows():
    if len(df_scan_exclude.loc[(df_scan_exclude.subID == row['subID'])&(df_scan_exclude.task==row['task'])])==0:
        df_participants_filter.append(pd.DataFrame(dict(zip(list(df_participants.columns), row.values)),index=[0]))
df_participants_filter = pd.concat(df_participants_filter,ignore_index=True)

print(f"After exclusion, we have {df_participants_filter.shape[0]} scans from {len(df_participants_filter.subID.unique())} participants in total.")
df_participants_filter
```

#### Excluding the stories that are not usable. 

```python
# Exclude the recited stories and the stories with special manipulation
List_story_exclude = ['merlin','shapesphysical','sherlock','notthefallshortscram','notthefalllongscram','schema']
List_condition_exclude = ['movie-physical-social','movie-social-physical','physical-movie-social','physical-social-movie']

df_participants_filter = df_participants_filter.loc[~df_participants_filter.task.isin(List_story_exclude),['subID','task','condition','N']]
df_participants_filter = df_participants_filter.loc[~df_participants_filter.condition.isin(List_condition_exclude)]
df_participants_filter['subID_task'] = df_participants_filter['subID']+'-'+df_participants_filter['task']
print(f'{len(df_participants_filter)} scans from {len(df_participants_filter.subID.unique())} participants left after excluding those stories.')
df_participants_filter.groupby(['task','condition']).sum().reset_index()
```

#### Reslice

```python
from nilearn import image
List_toReslice = []
for task in df_scans_filter.task.unique():
    df_tmp = df_scans_filter.loc[(df_scans_filter.task==task)&(~df_scans_filter.label.str.contains('confounds'))].reset_index(drop=True)
    [subID,folder,label,task,file] = df_tmp.loc[0,['subID','folder','label','task','files']]
    path_file = os.path.join(Dir_project,'derivatives',folder,subID,'func',file)
    img = image.load_img(path_file)
    affine = img.affine
    voxel_sizes = affine.diagonal()[:-1]
    if voxel_sizes[0]<3:
        List_toReslice.append(task)
        print(task,voxel_sizes)
```

```python
df_scans_filter.loc[(~df_scans_filter.task.isin(List_toReslice))&(~df_scans_filter.label.str.contains('confounds')),'img'] = df_scans_filter.loc[~df_scans_filter.task.isin(List_toReslice),'files']
df_scans_filter.loc[df_scans_filter.task.isin(List_toReslice)&(~df_scans_filter.label.str.contains('confounds')),'img'] = 'rs_'+df_scans_filter.loc[df_scans_filter.task.isin(List_toReslice),'files'] 
df_scans_filter
```
